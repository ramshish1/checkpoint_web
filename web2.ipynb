```python
#Cette fonction télécharge une page Web, vérifie qu'elle est accessible, 
# transforme son code HTML en un objet BeautifulSoup, puis renvoie cet objet afin que tu puisses facilement 
# extraire les informations de la page.
import requests 
# Importer BeautifulSoup de la bibliothèque bs4
# BeautifulSoup permet d'analyser le code HTML d'une page Web
from bs4 import BeautifulSoup
wiki_url = "https://fr.wikipedia.org/wiki/Astronomie"
def wiki_parser (wiki_url:str) :
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                    "AppleWebKit/537.36 (KHTML, like Gecko) "
                    "Chrome/114.0.0.0 Safari/537.36",
        "Accept-Language": "fr-FR,fr;q=0.9,en;q=0.8",
        "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8",
        "Connection": "keep-alive"
    }
    response = requests.get(wiki_url,headers = headers) 
    if response.status_code == 200 :
        # Analyser le contenu HTML de la page avec BeautifulSoup
        # 'html.parser' est le parseur HTML intégré à Python
        #parser:lit, analyse et organise un document (HTML, XML, JSON, etc.) afin que Python puisse le comprendre et le manipuler facileme
        soup = BeautifulSoup(response.content, 'html.parser')
        return soup
# soup = wiki_parser(wiki_url)    
# soup

```

```python
def get_article (soup):
    main_content = soup.find('div',id="bodyContent")
    all_content = soup.find_all(['p','h2','h3','h4'])
    article_content = {}
    current_h2 = None
    current_h3 = None
    current_h4 = None
    for element in all_content :

        if element.name == 'h2' : 
            current_h2 = element.get_text().strip()
            article_content[current_h2] = []

        elif element.name == 'h3':
            current_h3 = element.get_text().strip()
            article_content [current_h2].append({
                current_h3 :[]
            })
            current_h4 = None

        elif  element.name == 'h4':
            current_h4 =  element.get_text().strip()
            article_content[current_h2][-1][current_h3].append({
                current_h4 : []
            })
        
        else:
            paragraph = element.get_text().strip()
            if current_h4:
                article_content[current_h2][-1][current_h3][-1][current_h4].append(paragraph)

            if current_h3 and not current_h4:
                article_content[current_h2][-1][current_h3].append(paragraph)

            if current_h2 and not (current_h3 or current_h4):
                article_content[current_h2].append(paragraph)

    return article_content






```

```python
def get_links(soup):
    return ["https:"+link.get('href') for link in soup.find_all('a') if link.get("href") and link.get('href').startswith('//')]
```

```python
def wiki_scraper (url: str):
    soup = wiki_parser(url)
    if soup:
        article_title = get_title(soup)
        article_content = get_article(soup)
        article_links = get_links(soup)
        return article_title, article_content, article_links
```

```python
def display_article(article_title, article_content, article_links):
    print("-" * 100)
    print(article_title.center(100))
    print("-" * 100)

    for h2, h2_content in article_content.items():
        print(f"\n{h2}\n" + "=" * len(h2))

        for element in h2_content:
            if isinstance(element, str):
                # Paragraph directly under h2
                print(f"  {element}")

            elif isinstance(element, dict):
                for h3, h3_content in element.items():
                    print(f"\n  {h3}\n  " + "-" * len(h3))

                    for sub in h3_content:
                        if isinstance(sub, str):
                            # Paragraph directly under h3
                            print(f"    {sub}")

                        elif isinstance(sub, dict):
                            for h4, h4_content in sub.items():
                                print(f"\n    {h4}\n    " + "~" * len(h4))
                                for para in h4_content:
                                    print(f"      {para}")

    print("\n" + "-" * 100)
    print("External Links".center(100))
    print("-" * 100)
    for i, link in enumerate(article_links):
        print(f"{i+1}. {link}")
```

```python
title, content, links = wiki_scraper("https://fr.wikipedia.org/wiki/Apprentissage_automatique")
print(title)
print(content)
print(links)
```
